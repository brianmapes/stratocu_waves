```python
# ============================================================
# Categorical time-series forecast vs. persistence & climatology
# ============================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# ---------- 1. Load data ----------
# 3 columns per row: date, middle value (0-4, categorical), a 3rd column we ignore.
# No header assumed — change header=0 below if your file has one.
raw = pd.read_csv("/Users/bmapes/Downloads/data.csv", header=None, names=["date", "value", "col3"])

df = pd.DataFrame()
df["date"] = pd.to_datetime(raw["date"])
df["value"] = raw["value"].round().astype(int)   # 0..4 categorical
df = df.sort_values("date").reset_index(drop=True)

# Julian day = day-of-year (1-366), used later for the climatology baseline
df["doy"] = df["date"].dt.dayofyear

N_CLASSES = 5
MAX_LAG = 8     # forecast lead time: 1..8 steps ahead
N_HIST  = 8     # past observations fed to the model (t, t-1, ..., t-7)

# ---------- 2. Autoregressive feature set (same predictors for every horizon) ----------
for k in range(N_HIST):
    df[f"hist{k}"] = df["value"].shift(k)   # hist0 = value at t (now), hist7 = 7 steps back

hist_cols = [f"hist{k}" for k in range(N_HIST)]

# ---------- 3. Chronological train/test split ----------
split_idx = int(len(df) * 0.8)

# ---------- 4. Climatology model, built ONLY from the training period ----------
# For each julian day 1..366: frequency distribution of `value` in a 60-day
# window (±30 days, wrapping around the year boundary) centered on that day.
train_doy = df.loc[:split_idx - 1, "doy"].to_numpy()
train_val = df.loc[:split_idx - 1, "value"].to_numpy()

climatology = {}
for d in range(1, 367):
    diff = np.abs(train_doy - d)
    circ = np.minimum(diff, 366 - diff)          # wrap Dec <-> Jan
    window_vals = train_val[circ <= 30]
    counts = np.bincount(window_vals, minlength=N_CLASSES).astype(float)
    climatology[d] = counts / counts.sum() if counts.sum() > 0 else np.full(N_CLASSES, 1 / N_CLASSES)

rng = np.random.default_rng(42)
N_CLIM_DRAWS = 200   # Monte Carlo repeats, since climatology forecasts are random draws

# ---------- 5. Loop over forecast horizons 1..8, score each method ----------
records = []

for h in range(1, MAX_LAG + 1):
    d = df.copy()
    d["target"] = d["value"].shift(-h)        # value h steps ahead
    d["target_doy"] = d["doy"].shift(-h)       # calendar day of that target
    d = d.dropna(subset=hist_cols + ["target", "target_doy"]).reset_index(drop=True)

    s = int(len(d) * 0.8)
    train, test = d.iloc[:s], d.iloc[s:]

    X_train, y_train = train[hist_cols], train["target"].astype(int)
    X_test,  y_test  = test[hist_cols],  test["target"].astype(int)

    # --- Model: Random Forest classifier on the 8 lagged raw values ---
#    clf = RandomForestClassifier(n_estimators=300, max_depth=8, random_state=42)
    clf = RandomForestClassifier(n_estimators=300, max_depth=8, min_samples_leaf=20, random_state=42)
    clf.fit(X_train, y_train)
    model_acc = accuracy_score(y_test, clf.predict(X_test))

    # --- Persistence: forecast = today's value, unchanged ---
    persistence_acc = accuracy_score(y_test, test["hist0"].astype(int))

    # --- Climatology: random draw from the 60-day-window frequency table ---
    target_doys = test["target_doy"].astype(int).to_numpy()
    clim_accs = np.empty(N_CLIM_DRAWS)
    for r in range(N_CLIM_DRAWS):
        draws = np.array([rng.choice(N_CLASSES, p=climatology[dd]) for dd in target_doys])
        clim_accs[r] = accuracy_score(y_test, draws)
    climatology_acc = clim_accs.mean()

    records.append(dict(lag=h, model=model_acc, persistence=persistence_acc, climatology=climatology_acc))

results = pd.DataFrame(records).set_index("lag")
print(results.round(3))
```

```python
# ---------- 6. Plot: skill vs. forecast lag ----------
fig, ax = plt.subplots(figsize=(7, 4.5), facecolor="#fcfcfb")
ax.set_facecolor("#fcfcfb")

series = [
    ("model",       "Random Forest model",        "#2a78d6"),
    ("persistence", "Persistence",                 "#eb6834"),
    ("climatology", "Climatology (random draw)",   "#1baf7a"),
]
for col, label, color in series:
    ax.plot(results.index, results[col], marker="o", markersize=6,
             linewidth=2, color=color, label=label)

ax.set_xlabel("Forecast lag (steps ahead)")
ax.set_ylabel("Accuracy")
ax.set_xticks(results.index)
ax.set_ylim(0.18, 0.4)
ax.set_title("Categorical forecast skill vs. persistence and climatology")
ax.grid(True, color="#e1e0d9", linewidth=0.8)
for spine in ["top", "right"]:
    ax.spines[spine].set_visible(False)
ax.legend(frameon=False)
plt.tight_layout()
plt.show()
```
